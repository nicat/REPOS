# Task 1

```python
s=int(input('Enter your score: '))
if s>=91:
    print('A')
elif s>=81:
    print('B')
elif s>=71:
    print('C')
elif s>=61:
    print('D')
elif s>=51:
    print('E')
else:
    print('F')
```

# Task 2

```python
n=int(input('Enter number: '))
k=0
k1=0
for i in range(1,n+1):
    if i%15==0:
        k1+=1
        print('3,5')
    elif i%3==0:
        print('3')
        k1+=1
    elif i%5==0:
        print('5')
    else:
        print(i)
    k=k+i
print(k,k1)
```
